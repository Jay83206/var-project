```python
import pandas as pd
import numpy as np
import scipy.stats as stats
```

```python
data = pd.DataFrame(pd.read_csv("C:\\Users\\HP\\OneDrive\\Documents\\GitHub\\var-project\\data\\prices.csv"))
data = data.set_index('Date')
data.tail()
```

```python
returns = data.pct_change().dropna()
returns.tail()
```

```python
r = returns["HDFCBANK.NS"]
position = float(1_000_000)
threshold = np.percentile(r, 0.01*100)
hist_var = (-threshold) * position
No_of_time_threshold_breached_hist = float((r < threshold).sum())
print(f"Historical VaR = {hist_var:,.4f}, No_of_time_threshold_breached_hist = {No_of_time_threshold_breached_hist:,.2f}")
```

```python
mean = r.mean()
std_dev = r.std()
z_score = stats.norm.ppf(0.01)
threshold = mean + z_score * std_dev
para_var = (-threshold) * position
No_of_time_threshold_breached_para = float((r < threshold).sum())
print(f"Parametric VaR = {para_var:,.4f}, No_of_time_threshold_breached_para = {No_of_time_threshold_breached_para:,.2f}")
```

```python
prices = pd.read_csv("../data/prices.csv", index_col=0, parse_dates=True)
returns = prices.pct_change().dropna()
r = returns["HDFCBANK.NS"]

threshold = r.mean() + stats.norm.ppf(0.01) * r.std()
print("Threshold (should be a small negative number):", round(threshold, 4))

hits = r[r < threshold]
print("Breaches:", len(hits))
print(hits.index.strftime("%Y-%m").value_counts().sort_index())
print("Average return on breach days:", round(hits.mean(), 4))
```
